1. Pré-processamento dos dados e preparação do dataset.

```python
import pandas as pd

# Para acessar o arquivo, basta fazer o download dos microdados no seguinte endereço: https://download.inep.gov.br/microdados/microdados_enem_2024.zip
# O arquivo "RESULTADOS_2024.csv" encontra-se na pasta DADOS.

# Devido ao seu tamanho excessivamente grande, o GitHub não permite o armzenamento do arquivo no repositório;
# logo, deve-se fazer o download dos microdados e rodar localmente.

PATH_ENEM = "data/RESULTADOS_2024.csv"

# Carregar os dados
df_enem = pd.read_csv(
    PATH_ENEM,
    sep=";",
    encoding="latin1",
    low_memory=False
)

display(df_enem.head())
display(df_enem.info())
print("Dados carregados com sucesso!")
```

```python
# Excluir colunas irrelevantes.
df_enem = df_enem.drop(columns=[
                                'NU_SEQUENCIAL', 'NU_ANO', 'CO_MUNICIPIO_ESC', 'CO_UF_ESC', 'CO_MUNICIPIO_PROVA', 'CO_UF_PROVA',
                                'TP_DEPENDENCIA_ADM_ESC',	'TP_LOCALIZACAO_ESC', 'TP_SIT_FUNC_ESC', 'TP_PRESENCA_CN',
                                'TP_PRESENCA_CH',	'TP_PRESENCA_LC','TP_PRESENCA_MT',	'CO_PROVA_CN',	'CO_PROVA_CH',	'CO_PROVA_LC',
                                'CO_PROVA_MT', 'TX_RESPOSTAS_CN', 'TX_RESPOSTAS_CH', 'TX_RESPOSTAS_LC', 'TX_RESPOSTAS_MT',
                                'TP_LINGUA', 'TX_GABARITO_CN', 'TX_GABARITO_CH', 'TX_GABARITO_LC', 'TX_GABARITO_MT',
                                'TP_STATUS_REDACAO',	'NU_NOTA_COMP1',	'NU_NOTA_COMP2',	'NU_NOTA_COMP3',	'NU_NOTA_COMP4',	'NU_NOTA_COMP5'
                                ]
                       )
display(df_enem.head())
print("Colunas excluídas com sucesso!")
```

```python
# Renomear as colunas, para tornar a análise mais intuitiva.
df_enem = df_enem.rename(columns=
    {
    "CO_ESCOLA": "codigo_escola",
    "NO_MUNICIPIO_ESC": "municipio_escola",
    "SG_UF_ESC": "uf_escola",
    "NO_MUNICIPIO_PROVA": "municipio_prova",
    "SG_UF_PROVA": "uf_prova",
    "NU_NOTA_CN": "nota_ciencias_natureza",
    "NU_NOTA_CH": "nota_ciencias_humanas",
    "NU_NOTA_LC": "nota_linguagens_codigos",
    "NU_NOTA_MT": "nota_matematica",
    "NU_NOTA_REDACAO": "nota_redacao"
    }
                         )

display(df_enem.head())
print("Colunas renomeadas com sucesso!")
```

```python
# Converter o tipo da coluna 'codigo_escola' para 'Int64'.
df_enem["codigo_escola"] = df_enem["codigo_escola"].astype('Int64')

display(df_enem.info())
print("Tipo da coluna 'codigo_escola' convertido com sucesso!")
```

```python
# Filtrar SP nas colunas 'uf_escola' e 'uf_prova'
df_enem_sp_escola = df_enem[df_enem["uf_escola"] == "SP"].copy()
df_enem_sp_prova = df_enem[df_enem["uf_prova"] == "SP"].copy()
```

```python
display(df_enem_sp_escola.info())

print("""Com o filtro pela coluna da escola, são retornadas 285199 linhas,
correspondentes aos candidatos concluintes do Ensino Médio em escolas paulistas.""")
```

```python
display(df_enem_sp_prova.info())

print("""Com o filtro pela coluna da prova, são retornadas 647215 linhas, correspondentes a todos os candidatos - não considerando faixa etária
ou vinculação formal com escola, por exemplo - que realizaram as provas em cidades do estado de São Paulo.""")

print("""Uma vez que a amplitude de faixas etárias pode representar diferentes realidades sociais - de tentativas de ingresso tardio, devido à
diferentes razões, à postulantes à segunda graduação - e como o intuito último da presente pesquisa é a construção de um modelo inicial
que possa embasar políticas públicas a partir das análises, optou-se por delimitá-la apenas aos concluintes do Ensino Médio, por entender
que a faixa etária fornece um solo comum mais homogêneo. Nada impede, contudo, que retomemos esse modelo inicial posteriormente, alterando
o foco da abordagem. Assim, nas próximas células, será utilizado apenas o DataFrame 'df_enem_sp_escola'.""")
```

```python
# Reorganizar as colunas.
# Excluir colunas obsoletas: 'uf_escola', 'municipio_prova' e 'uf_prova'.
df_enem_sp_escola = df_enem_sp_escola.drop(columns=['uf_escola', 'municipio_prova', 'uf_prova'])

# Renomear a coluna 'municipio_escola' para 'municipio', simplificando a compreensão.
df_enem_sp_escola = df_enem_sp_escola.rename(columns={'municipio_escola': 'municipio'})

display(df_enem_sp_escola.head())
print("Colunas reorganizadas com sucesso!")
```

```python
# Criar média das notas das provas objetivas - já que a nota da prova de redação, dada a sua natureza particular, deveria ser analisada de maneira isolada.
df_enem_sp_escola["media_notas_objetivas"] = df_enem_sp_escola[
    ["nota_ciencias_natureza", "nota_ciencias_humanas", "nota_linguagens_codigos", "nota_matematica"]
].mean(axis=1)

display(df_enem_sp_escola.head())
print("Coluna 'media_notas_objetivas' criada com sucesso!")
```

```python
# Arredondar a coluna 'media_notas_objetivas' para uma casa decimal.
df_enem_sp_escola['media_notas_objetivas'] = df_enem_sp_escola['media_notas_objetivas'].round(1)

display(df_enem_sp_escola.head())
print("Coluna 'media_notas_objetivas' arredondada para uma casa decimal com sucesso!")
```

```python
# Remover candidatos ausentes.
# Candidatos ausentes nos 2 dias.
df_enem_sp_escola = df_enem_sp_escola[df_enem_sp_escola["media_notas_objetivas"].notna()]
# Candidatos ausentes no 1o dia (Exatas).
df_enem_sp_escola = df_enem_sp_escola[df_enem_sp_escola["nota_matematica"].notna()]
# Candidatos ausentes no 2o dia (Humanas).
df_enem_sp_escola = df_enem_sp_escola[df_enem_sp_escola["nota_ciencias_humanas"].notna()]

display(df_enem_sp_escola.info())
print("Candidatos ausentes excluídos com sucesso!")
```

```python
# Para acessar o arquivo com a lista das escolas de acordo com o 'Código INEP', basta acessar o seguinte endereço: http://anonymousdata.inep.gov.br/analytics/saw.dll?Portal.
# Uma vez nele, basta aplicar o filtro desejado - em nosso caso, UF == SP - e exportar para um arquivo .csv.

# Como o arquivo é pequeno, mantemos uma cópia no repositório do GitHub.

PATH_ESCOLAS = "data/lista_escolas.csv"

# Carregar os dados
df_escolas = pd.read_csv(
    PATH_ESCOLAS,
    sep=",",
    encoding="utf-8",
    low_memory=False
)

display(df_escolas.info())
print("Dados carregados com sucesso!")
```

```python
# Converter a coluna 'Código INEP' em df_escolas para o tipo Int64 para compatibilidade com 'codigo_escola'.
df_escolas['Código INEP'] = df_escolas['Código INEP'].astype('Int64')

display(df_escolas.info())
print("Tipo da coluna 'Código INEP' convertido com sucesso!")

# Realizar o merge para adicionar as colunas com os detalhes das escolas.
df_enem_sp_escola = pd.merge(
    df_enem_sp_escola,
    df_escolas[['Código INEP', 'Categoria Administrativa', 'Dependência Administrativa', 'Categoria Escola Privada']],
    left_on='codigo_escola',
    right_on='Código INEP',
    how='left'
)

# Excluir as colunas 'Código INEP', redundante do merge, e 'codigo_escola', doravante desnecessária.
df_enem_sp_escola = df_enem_sp_escola.drop(columns=['Código INEP', 'codigo_escola'])

display(df_enem_sp_escola.head())
print("Colunas sobre as escolas adicionadas com sucesso!")
```

```python
# Renomear novas colunas.
df_enem_sp_escola = df_enem_sp_escola.rename(columns=
    {
     "Categoria Administrativa": "cat_adm",
     "Dependência Administrativa": "dep_adm",
     "Categoria Escola Privada": "cat_esc_priv"
    }
                                             )

display(df_enem_sp_escola.head())
print("Colunas renomeadas com sucesso!")
```

```python
# Identificar linhas onde 'cat_adm' é nula.
linhas_nulas_cat_adm = df_enem_sp_escola[df_enem_sp_escola['cat_adm'].isnull()]

print("Linhas onde 'cat_adm' é nula (primeiras 5):")
display(linhas_nulas_cat_adm.head())
print(f"Total de linhas nulas na coluna 'cat_adm': {len(linhas_nulas_cat_adm)}")
```

```python
# Excluir as linhas em que o código da escola não corresponde ao código da base do INEP da escola, pois essa imprecisão pode distorcer os dados.
df_enem_sp_escola = df_enem_sp_escola.dropna(subset=['cat_adm'])

display(df_enem_sp_escola.info())
print("Linhas com 'cat_adm' nulas excluídas com sucesso!")
```

```python
# Exportar 'df_enem_sp_escola'
df_enem_sp_escola.to_csv('data/df_enem_sp.csv', index=False)
print("DataFrame 'df_enem_sp_escola.csv' exportado com sucesso para 'data/df_enem_sp.csv'.")
```

```python
# Agrupar os dados por município.
base_municipio = (
    df_enem_sp_escola.groupby("municipio")
    .agg(
        media_notas=("media_notas_objetivas", "mean"),
        total_participantes=("media_notas_objetivas", "size"),
        desvio_notas=("media_notas_objetivas", "std")
    )
    .reset_index()
)

# Arredondar as colunas 'media_notas' e 'desvio_notas' para uma casa decimal
base_municipio['media_notas'] = base_municipio['media_notas'].round(1)
base_municipio['desvio_notas'] = base_municipio['desvio_notas'].round(1)

display(base_municipio.head())
print("DataFrame 'base_municipio' criado com sucesso!")
```

```python
# Será utilizada apenas uma amostra, com os 200 municípios com mais participantes na prova, para alimentar os algoritmos de regressão.
# Isso posto, ordena-se os municípios pela coluna 'total_participantes' de maneira decrescente.
base_municipio = base_municipio.sort_values(by='total_participantes', ascending=False).reset_index(drop=True)

# Selecionar os 200 primeiros municípios, que serão a base para a constituição do arquivo 'censo_2022.csv'.
base_municipio_top200 = base_municipio['municipio'].head(200).copy()
print("DataFrame 'base_municipio_top200' criado com sucesso!")

print("Os 200 primeiros municípios com mais participantes são:")
display(base_municipio_top200.head(200))

# Exportar a lista dos 200 primeiros municípios.
# base_municipio_top200.to_csv('data/base_municipio_top200.csv', index=False)
# print("DataFrame 'base_municipio_top200' exportado com sucesso para 'data/base_municipio_top200.csv'.")
print("A linha que exporta o DataFrame foi comentada, uma vez que o arquivo é temporário.")
```

```python
print("""Em respeito à LGPD, o INEP dividiu os microdados da edição de 2024 em dois datasets: um com os dados socioeconômicos indivduais dos participantes
e um com os resultados das provas, sem chave de conexão entre eles. Isso limita bastante a pesquisa aplicada: foi preciso, por exemplo,
trabalhar apenas com variáveis a nível municipal, extraídas do Censo 2022, realizado pelo IBGE, que ainda permitem investigar desigualdades
territoriais de forma robusta, evitando inferências indevidas a partir de dados não vinculáveis. Contudo, como veremos mais adiante, a perspectiva individual
certamente ajudaria nas predições.""")

# O IPDM - Índice Paulista de Desenvolvimento Municipal - pode ser consultado no seguinte endereço:
# https://repositorio.seade.gov.br/dataset/2be01068-49cc-4cc1-85b8-38bd867aefc4/resource/887a7554-57cd-4a48-8c61-e9ea567b4bfb/download/ipdm.zip
# Os demais dados foram compilados à partir dos indicadores presentes nos seguintes painéis: https://censo2022.ibge.gov.br/panorama/indicadores.html?localidade=3550308&tema=5
# e https://censo2022.ibge.gov.br/panorama/indicadores.html?localidade=3550308&tema=11

# Como o tamanho do arquivo é irrisório, mantemos uma cópia no repositório do GitHub.

PATH_CENSO = "data/censo_2022.csv"

# Carregar os dados
df_censo = pd.read_csv(
    PATH_CENSO,
    sep=",",
    encoding="utf-8",
    low_memory=False
)
display(df_censo.head())
display(df_censo.info())
print("Dados carregados com sucesso!")
```

```python
# Arredondar a coluna 'taxa_alfabetizacao' para três casas decimais
df_censo['taxa_alfabetizacao'] = df_censo['taxa_alfabetizacao'].round(3)

display(df_censo.head())
print("Coluna 'taxa_alfabetizacao' arredondada para três casas decimais com sucesso!")
```

```python
# Realizar o merge dos DataFrames 'base_municipio' e 'df_censo' segundo a chave 'municipio'
df_final = pd.merge(base_municipio, df_censo, on='municipio', how='inner')

# Exibir as primeiras linhas do DataFrame final e suas informações
display(df_final.head())
display(df_final.info())

# Exportar o 'df_final'
df_final.to_csv('data/df_final.csv', index=False)
print("DataFrame 'df_final.csv' exportado com sucesso para 'data/df_final.csv'.")
```
